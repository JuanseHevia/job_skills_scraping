```python
import os
import dotenv

dotenv.load_dotenv(dotenv.find_dotenv())
print("Variables de entorno cargadas desde el archivo .env")
```

```python
URLS_TO_SCRAPE = "scraped/test_austral_20250609_0946-SHORT.txt"
SAVEDIR = "data/test_austral/test.json"

# si el base path no existe, lo creamos
if not os.path.exists(os.path.dirname(SAVEDIR)):
    os.makedirs(os.path.dirname(SAVEDIR))
    print(f"Directorio creado: {os.path.dirname(SAVEDIR)}")

assert os.path.exists(URLS_TO_SCRAPE), f"El archivo {URLS_TO_SCRAPE} no existe."

print("Vamos a cargar URLs desde el archivo:", URLS_TO_SCRAPE)
```

```python
# Exploremos qué contiene el archivo de URLs
with open(URLS_TO_SCRAPE, "r") as file:
    urls = file.readlines()
urls = [url.strip() for url in urls if url.strip()]  # Limpiar espacios en blanco y eliminar líneas vacías
print(f"Se encontraron {len(urls)} URLs en el archivo.")

# veamos las primeras 5 URLs
print("Primeras 5 URLs:")
for url in urls[:5]:
    print(url)
```

### Llamamos al script de procesamiento vía OpenAI

```python
import subprocess

result = subprocess.run(
    ["uv", "run" ,"get_job_data.py", "--input_path", URLS_TO_SCRAPE, "--output_path", SAVEDIR],
    capture_output=True,
    text=True
)

print(result.stdout)
if result.stderr:
    print("Error:", result.stderr)
```
